```python
import soundfile as sf
import io

audio_array, sampling_rate = sf.read(
    io.BytesIO(example["audio"]["bytes"])
)

predictions = classifier(audio_array)

print(predictions)
```

```python
import numpy as np

correct = 0
total = 10

for i in range(total):
    example = minds.cast_column("audio", Audio(decode=False))[i]

    audio_array, sampling_rate = sf.read(
        io.BytesIO(example["audio"]["bytes"])
    )

    predictions = classifier({
        "raw": audio_array,
        "sampling_rate": sampling_rate
    })

    predicted_label = predictions[0]["label"]
    actual_label = minds.features["intent_class"].int2str(
        example["intent_class"]
    )

    if predicted_label == actual_label:
        correct += 1

    print(f"Sample {i+1}")
    print(f"Actual: {actual_label}")
    print(f"Predicted: {predicted_label}")
    print(f"Confidence: {predictions[0]['score']:.2%}")
    print("-" * 30)

print(f"Accuracy: {correct}/{total} = {correct/total:.2%}")
```
